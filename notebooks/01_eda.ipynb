# EDA: NSL-KDD Network Traffic

Quick exploratory analysis before training.

```python
import sys
sys.path.append("..")

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_nsl_kdd, NSL_KDD_COLUMNS

sns.set_theme(style="whitegrid")
```

```python
# Adjust paths after downloading NSL-KDD into data/
train_df, test_df = load_nsl_kdd(
    "../data/KDDTrain+.txt",
    "../data/KDDTest+.txt",
)
print(train_df.shape, test_df.shape)
train_df.head()
```

```python
# Class balance
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
train_df["target"].value_counts().plot(kind="bar", ax=ax[0], title="Train")
test_df["target"].value_counts().plot(kind="bar", ax=ax[1], title="Test")
plt.tight_layout()
plt.show()
```

```python
# Top attack types in original labels
print(train_df["label"].value_counts().head(15))
```

```python
# Protocol distribution
pd.crosstab(train_df["protocol_type"], train_df["target"]).plot(kind="bar", stacked=True)
plt.title("Protocol vs Target")
plt.show()
```
